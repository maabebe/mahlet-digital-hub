# Section 5: Data Science 

Some broader questions that I have about the DWTS dataset is whether there is a correlation between age and the final 4 contestants on DWTS, as well as the link between age and the first person eliminated each season. Therefore, I will take two samples from my dataset, one with the final 4 contestants on each season, as well as the bottom 4 contestants per season to see if there is a correlation. Then, I will use Python to conduct a hypothesis test on each of these samples to see whether my inital hypothesis is correct. This is because I believe that looking at exclusively winners won't give us an accurate depiction of the correlation between age and final placement on DWTS.

# TOP PLACEMENTS

```python
dwts_question = "How does age affect your final placement on DWTS?"
dwts_question
```

Null Hypothesis: μ = 39 years old 

Alternative Hypothesis: μ < 39 years old 

```python
top_placements = df_dwts[df_dwts.placement <= 4.0]
top_placements
```

```python
top_placements_mean = top_placements["celebrity_age_during_season"].mean()
top_placements_mean
```

```python
top_placements_std = top_placements["celebrity_age_during_season"].std()
top_placements_std
```

```python
top_placements_samplesize = top_placements["celebrity_age_during_season"].count()
top_placements_samplesize
```

```python
top_placements_standarderror = (top_placements_std)/(top_placements_samplesize)**(1/2)
top_placements_standarderror
```

```python
top_placements_z_stat = (top_placements_mean - 39)/top_placements_standarderror
top_placements_z_stat
```

```python
from statsmodels.stats.weightstats import ztest 
```

```python
top_placements_results = ztest(top_placements["celebrity_age_during_season"].dropna(),value=39,alternative="smaller")
top_placements_results 
```

Our p-value is approximently 2.42 * e^-22. Therefore, we reject the null hypothesis in favor of our alternative hypothesis as our p-value is less than our significance level of 0.05. Based on this outcome, there is evidence to suggest that the average age of the final 4 contestnts on each season of DWTS is less than 39 years old. 

```python
tp_mean = top_placements["celebrity_age_during_season"].mean()
tp_mean
```

# BOTTOM PLACEMENTS

Null Hypothesis: μ = 39 years old 

Alternative Hypothesis: μ > 39 years old 

```python
bottom_placements = df_dwts[df_dwts.placement >= 12.0]
bottom_placements
```

```python
bottom_placements_std = bottom_placements["celebrity_age_during_season"].std()
bottom_placements_std
```

```python
bottom_placements_samplesize = bottom_placements["celebrity_age_during_season"].count()
bottom_placements_samplesize
```

```python
bottom_placements_standarderror = (bottom_placements_std)/(bottom_placements_samplesize)**(1/2)
bottom_placements_standarderror
```

```python
bottom_placements_z_stat = (bottom_placements_mean - 39)/bottom_placements_standarderror
bottom_placements_z_stat
```

```python
from statsmodels.stats.weightstats import ztest 
```

```python
bottom_placements_results = ztest(bottom_placements["celebrity_age_during_season"].dropna(),value=39,alternative="larger")
bottom_placements_results 
```

Our p-value is approximently 2.81 * e^8. Therefore, we reject the null hypothesis in favor of our alternative hypothesis as our p-value is less than our significance level of 0.05. Based on this outcome, there is evidence to suggest that the average age of the bottom 4 placements is greater than 39 years old. 

```python
bottom_placements_mean = bottom_placements["celebrity_age_during_season"].mean()
bottom_placements_mean
```
